```python
import numpy as np
import matplotlib.pyplot as plt
from tensorboard.backend.event_processing import event_accumulator

save_path = "E:\\python_work\\other\\0\\events.out.tfevents.1776325797.autodl-container-3a9849955e-788f4293.1698.0"
ea = event_accumulator.EventAccumulator(
    save_path,
    size_guidance={
        event_accumulator.SCALARS: 0,  # 0 表示加载全部
    }
)

ea.Reload()
print("Available tags:", ea.Tags())
def get_scalar(tag):
    events = ea.Scalars(tag)
    steps = np.array([e.step for e in events])
    values = np.array([e.value for e in events])
    return steps, values

step, skip_res = get_scalar("train/skip_loss")
step, train_loss = get_scalar("train/train_loss")
step,  up_loss = get_scalar("train/up_loss")

step, psnr_values = get_scalar("val/psnr")
step, ssim_values = get_scalar("val/ssim")

idx_max_psnr = np.argmax(psnr_values)
val_max_psnr = psnr_values[idx_max_psnr]
step_max_psnr = step[idx_max_psnr]

# print(plt.rcParams.keys())

def apply_custom_style():
    # 1. 基础皮肤
    # plt.style.use('seaborn-v0_8-darkgrid')
    
    # 2. 自定义细节参数
    custom_params = {
        "axes.facecolor": "#eaf4f8",      # 浅色背景（如需纯深色可改 #1a1a2e）
        "axes.grid": True,
        "grid.color": "#000000",
        "grid.linestyle": "--",
        "grid.alpha": 0.5,
        "axes.edgecolor": "#8A8A8A",
        "figure.facecolor": "#FFFFFFFF",      # 外部画布颜色
        "font.sans-serif": ["Arial", "Liberation Sans", "DejaVu Sans"],
        "axes.titlesize": 14,
        "axes.titleweight": "bold",
    }
    plt.rcParams.update(custom_params)

apply_custom_style()
fig, ax1 = plt.subplots(figsize=(8, 5), dpi=120)
line_style1 = {"linewidth": 1.5, "alpha": 1.0, "linestyle": "-"}
line_style2 = {"linewidth": 1.3, "alpha": 0.8, "linestyle": "--", "color": "tab:red"}
ax1.plot(step, skip_res, label='skip_loss', **line_style1)
ax1.plot(step, train_loss, label='train_loss', **line_style1)
ax1.plot(step, up_loss, label='up_loss', **line_style1)
ax1.set_yscale('log')
ax1.set_xlabel("Step", fontsize=11)
ax1.set_ylabel("Loss (log scale)", fontsize=11)
ax1.spines['top'].set_visible(False)
ax2 = ax1.twinx()
ax2.plot(step, psnr_values, label='PSNR', **line_style2)
# ax2.plot(step, ssim_values, label='SSIM', **line_style2)
ax2.set_ylabel("Metric Value", fontsize=11)
ax2.spines['top'].set_visible(False)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    loc='upper center',
    bbox_to_anchor=(0.5, 1.15),
    ncol=3,
    frameon=False
)

arrow_fmt = dict(facecolor='black', edgecolor='black', arrowstyle='->', connectionstyle='arc3,rad=0.2', linewidth=1)
text_fmt = dict(fontsize=9, fontweight='bold', bbox=dict(boxstyle='round,pad=0.3', fc='white', ec='gray', alpha=0.8))

ax2.annotate(f'Max PSNR: {val_max_psnr:.2f}\n(Step {step_max_psnr})',
            xy=(step_max_psnr, val_max_psnr),  # 被箭头发出的点 (数据坐标)
            xytext=(step_max_psnr - 50, val_max_psnr - 5.0), # 文字所在的位置
            textcoords='data',
            arrowprops=dict(arrowstyle='-|>', connectionstyle='arc3,rad=-0.1', color="#000000", lw=0.8),
            color="#AB3E3E", fontsize=9, fontweight='bold',
            horizontalalignment='right')

ax2.scatter(step_max_psnr, val_max_psnr, 
            s=60,                # 点的大小
            c='#ff1744',         # 点的填充颜色（如：鲜红色）
            edgecolors='white',  # 点的边框颜色（白色能让点在深色背景中更跳跃）
            linewidths=1.5,      # 边框宽度
            zorder=5)            # 确保点在曲线（zorder通常为2）的上方

plt.tight_layout()
plt.show()  
```
